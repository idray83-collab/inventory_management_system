## Sports Player Performance Analysis

```python
goals_scored = [12, 7, 9, 15, 6, 10] #this is records of goals scored

print(goals_scored)
```

```python
total_goals_scored = sum(goals_scored) #this is the total number of goals scored

print(total_goals_scored)
```

```python
difference = max(goals_scored) - min(goals_scored) #difference between the highest and the lowest

print(difference)
```

```python
average_goals_scored = total_goals_scored / len(goals_scored) #average goals per match

print(average_goals_scored)
```

```python
goals_scored.append(11) #11 goals added for a new match

print(goals_scored)
```

```python
goals_scored.remove(min(goals_scored)) #this removes lowest-scoring match
                    
print(goals_scored)
```

```python
print(goals_scored) #final updated performance data
```

## CPU Temperature Monitoring System

```python
temperature = (45, 50, 55, 48, 52) #cpu temperature readings

print(temperature)
```

```python
temperature[0] #first temperature reading
```

```python
temperature[-1] #last temperature reading
```

```python
min(temperature)
```

```python
max(temperature)
```

```python
difference = max(temperature) - min(temperature) #this finds difference between the highest and lowest temperature

print(difference)
```

```python
total_temperature = sum(temperature)

print(total_temperature)
```

```python
average_temperature = total_temperature / len(temperature) #average cpu temperature

print(average_temperature)
```

Tuple is more suitable than a list for storing temperature data because of the following:

- Tuple is immutable. CPU temperature readings collected at a specific time should usually remain permanent for accurate monitoring and analysis.

- Tuple uses less memory. It consumes less memory than lists, making it efficient for storing constant sensor data.
